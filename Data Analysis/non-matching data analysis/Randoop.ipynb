```python
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import numpy as np
```

```python
# Load the datasets
randoop_errors = pd.read_parquet("/Users/bilegjargalbattulga/Desktop/Master Thesis/data analysis/randoop_errors.parquet")
merged_randoop_errors = pd.read_parquet("/Users/bilegjargalbattulga/Desktop/Master Thesis/data analysis/merged_randoop_errors.parquet")
sorl_df = pd.read_parquet('/Users/bilegjargalbattulga/Desktop/Final/sorl_df.parquet')
```

```python
# Count unique values in key columns
unique_counts = merged_randoop_errors[['ABSTRACTIONID', 'SHEETID', 'SYSTEMID']].nunique()
print("\nUnique Counts:")
print(unique_counts)
```

```python
#  Analyze distinct RAWVALUE counts
outputs = merged_randoop_errors.query("TYPE == 'value'")
rawvalue_counts = outputs['RAWVALUE'].value_counts().reset_index()
rawvalue_counts.columns = ['RawValue', 'Occurrences']
print("\nRawValue Counts:")
print(rawvalue_counts)
```

```python
# Filter ABSTRACTIONIDs with only one SYSTEMID
systemid_count = merged_randoop_errors.groupby('ABSTRACTIONID')['SYSTEMID'].nunique().reset_index()
systemid_count.columns = ['ABSTRACTIONID', 'Unique_SYSTEMID_Count']
single_systemid_abstraction_ids = systemid_count[systemid_count['Unique_SYSTEMID_Count'] == 1]['ABSTRACTIONID']
single_systemid_df = merged_randoop_errors[merged_randoop_errors['ABSTRACTIONID'].isin(single_systemid_abstraction_ids)]
result = single_systemid_df[['ABSTRACTIONID', 'SYSTEMID']].drop_duplicates()
print("\nABSTRACTIONIDs with Single SYSTEMID:")
print(result)
```

```python
# Sort by Unique_SYSTEMID_Count in descending order
sorted_systemid_count = systemid_count.sort_values(by='Unique_SYSTEMID_Count', ascending=False)
print("\nSorted SystemID Counts:")
print(sorted_systemid_count)
```

```python
# Count unique input types per ABSTRACTIONID
inputs = merged_randoop_errors.query("TYPE == 'input_value'")
input_type_counts = inputs.groupby('ABSTRACTIONID').agg({'VALUE': 'nunique'}).reset_index()
input_type_counts.columns = ['AbstractionID', 'Number of Input Types']
input_type_counts = input_type_counts.sort_values(by='Number of Input Types', ascending=True)

print("\nInput Type Counts:")
print(input_type_counts)
```

```python
import plotly.graph_objects as go

# Merge systemid_count and input_type_counts for visualization
# Ensure correct column names before merging
systemid_count.columns = ['ABSTRACTIONID', 'Unique_SYSTEMID_Count']
input_type_counts.columns = ['ABSTRACTIONID', 'Number of Input Types']

merged_counts = systemid_count.merge(
    input_type_counts, on='ABSTRACTIONID', how='inner'
)

# Sort by Unique SYSTEMID Count in descending order
merged_counts = merged_counts.sort_values(by='Unique_SYSTEMID_Count', ascending=False)

# Set fixed chart height
chart_height = 3500  # Fixed height as per your preference

# Create a dual-axis horizontal chart
fig = go.Figure()

# Horizontal bar chart for Unique SYSTEMID Count
fig.add_trace(
    go.Bar(
        y=merged_counts['ABSTRACTIONID'],  # ABSTRACTIONIDs on y-axis
        x=merged_counts['Unique_SYSTEMID_Count'],
        name='Unique SYSTEMID Count',
        orientation='h',  # Horizontal bars
        marker=dict(color='blue'),
        hovertemplate='<b>ABSTRACTIONID:</b> %{y}<br><b>Unique SYSTEMID Count:</b> %{x}<extra></extra>'
    )
)

# Horizontal line chart for Number of Input Types
fig.add_trace(
    go.Scatter(
        y=merged_counts['ABSTRACTIONID'],  # ABSTRACTIONIDs on y-axis
        x=merged_counts['Number of Input Types'],
        name='Number of Input Types',
        mode='lines+markers',  # Line with markers
        marker=dict(color='orange', size=6, symbol='circle'),  # Orange markers
        line=dict(width=2, shape='linear'),  # Straight lines between points
        hovertemplate='<b>ABSTRACTIONID:</b> %{y}<br><b>Number of Input Types:</b> %{x}<extra></extra>'
    )
)

# Update layout to remove unnecessary space and improve readability
fig.update_layout(
    yaxis=dict(
        title="ABSTRACTIONID",  # ABSTRACTIONIDs on y-axis
        showticklabels=True,
        tickmode='linear',  # Ensure all tick labels are shown
        automargin=True,  # Automatically adjust margins
        range=[-0.5, len(merged_counts) - 0.5],  # Remove extra space at top and bottom
    ),
    xaxis=dict(
        title="Count",
        side='bottom',
        showgrid=True,
    ),
    legend=dict(x=0.8, y=1.02, orientation="h"),  # Position legend closer to the graph
    barmode='overlay',
    template='plotly_white',
    height=chart_height,  # Set fixed height
    width=1200,  # Adjust width for better readability
    margin=dict(l=200, r=20, t=50, b=20),  # Fine-tuned margins to remove space
)

# Show the plot
fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Filter data into three categories
one_systemid = merged_counts[merged_counts['Unique_SYSTEMID_Count'] == 1]
low_occurrences = merged_counts[
    (merged_counts['Unique_SYSTEMID_Count'] > 1) & (merged_counts['Unique_SYSTEMID_Count'] <= 10)
]
high_occurrences = merged_counts[merged_counts['Unique_SYSTEMID_Count'] > 10]

# Calculate counts for each group
count_one_systemid = len(one_systemid)
count_low_occurrences = len(low_occurrences)
count_high_occurrences = len(high_occurrences)

# Create a grid layout with 1 row and 3 columns
fig = make_subplots(
    rows=1, cols=3,
    subplot_titles=[
        f"1 SYSTEMID ({count_one_systemid} Abstractions)",
        f"Low Occurrences (2 ≤ SYSTEMID Count ≤ 10) ({count_low_occurrences} Abstractions)",
        f"High Occurrences (SYSTEMID Count > 10) ({count_high_occurrences} Abstractions)"
    ],
    horizontal_spacing=0.1  # Space between columns
)

# Function to add traces to a subplot
def add_traces(fig, data, col, bar_color, line_color, name_suffix):
    # Bar chart for Unique SYSTEMID Count
    fig.add_trace(
        go.Bar(
            y=data['ABSTRACTIONID'],
            x=data['Unique_SYSTEMID_Count'],
            name=f"Unique SYSTEMID Count ({name_suffix})",
            orientation='h',
            marker=dict(color=bar_color),
            hovertemplate='%{x}'  # Only show the x value (Unique SYSTEMID Count)
        ),
        row=1, col=col
    )

    # Line chart for Number of Input Types
    fig.add_trace(
        go.Scatter(
            y=data['ABSTRACTIONID'],
            x=data['Number of Input Types'],
            name=f"Number of Input Types ({name_suffix})",
            mode='lines+markers',
            marker=dict(color=line_color, size=6),
            hovertemplate='%{x}'  # Only show the x value (Number of Input Types)
        ),
        row=1, col=col
    )

# Add data for each category in the grid layout
add_traces(fig, one_systemid, col=1, bar_color='blue', line_color='orange', name_suffix="1 SYSTEMID")
add_traces(fig, low_occurrences, col=2, bar_color='green', line_color='red', name_suffix="Low Occurrences")
add_traces(fig, high_occurrences, col=3, bar_color='purple', line_color='cyan', name_suffix="High Occurrences")

# Update layout
fig.update_layout(
    height=1000,  # Total figure height
    width=1800,   # Adjust width for the grid
    title="Analysis of SYSTEMID and Input Types Across Abstraction IDs",
    showlegend=False,  # You can enable this if needed
    template='plotly_white',
    margin=dict(l=200, r=20, t=50, b=20)  # Margins around the grid
)

# Show the grid layout
fig.show()
```
